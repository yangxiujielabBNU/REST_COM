# BrainVision 数据转换为 MNE .fif 格式

将 DATA 目录下的 BrainVision 格式数据转换为 MNE 的 .fif 格式

```python
# 导入必要的库
import os
import mne
```

```python
# 设置路径和参数
data_path = r'd:\LYW\REST_COM\DATA\\正常被试\\'
save_path = r'.\data\1raw_rename\\'
sub_id = '080'  # 测试被试
```

```python
# 创建输出目录
os.makedirs(save_path, exist_ok=True)
print(f'输出目录: {save_path}')
```

```python
# 读取 BrainVision 数据
fname_original = f'{data_path}{sub_id}\\{sub_id} REST.vhdr'
print(f'正在读取: {fname_original}')
raw = mne.io.read_raw_brainvision(fname_original, preload=True, verbose=None)

# 显示数据信息
print(f'\n数据信息:')
print(f'  通道数: {len(raw.ch_names)}')
print(f'  采样率: {raw.info["sfreq"]} Hz')
print(f'  时长: {raw.times[-1]:.2f} 秒')
```

```python
# 保存为 .fif 格式（使用 MNE 命名规范）
fname_new = f'{save_path}{sub_id}_raw.fif'
print(f'\n正在保存到: {fname_new}')
raw.save(fname_new, overwrite=True)
print(f'✓ 成功转换被试 {sub_id}')
```

```python
# 验证转换后的文件
print(f'\n验证文件...')
raw_loaded = mne.io.read_raw_fif(fname_new, preload=False, verbose=None)
print(f'✓ 文件验证成功！')
print(f'  验证通道数: {len(raw_loaded.ch_names)}')
print(f'  验证采样率: {raw_loaded.info["sfreq"]} Hz')
```

## 批量处理所有被试

下面的代码会自动处理所有 12 个被试

```python
# 批量处理所有被试
sub_list = ['001', '021', '022', '024', '047', '049', '051', '071', '072', '073', '076', '092']

print(f'开始批量处理 {len(sub_list)} 个被试...\n')

success_count = 0
failed_list = []

for sub_id in sub_list:
    try:
        # 读取数据
        fname_original = f'{data_path}{sub_id}\\{sub_id} REST.vhdr'
        print(f'[{sub_id}] 正在读取...')
        raw = mne.io.read_raw_brainvision(fname_original, preload=True, verbose=None)
        
        # 保存为 .fif 格式
        fname_new = f'{save_path}{sub_id}_raw.fif'
        raw.save(fname_new, overwrite=True)
        
        print(f'[{sub_id}] ✓ 转换成功 ({len(raw.ch_names)} 通道, {raw.info["sfreq"]} Hz)\n')
        success_count += 1
        
    except Exception as e:
        print(f'[{sub_id}] ✗ 转换失败: {str(e)}\n')
        failed_list.append(sub_id)
```

```python
# 显示处理结果摘要
print('='*50)
print(f'批量处理完成！')
print(f'成功: {success_count}/{len(sub_list)}')
if failed_list:
    print(f'失败的被试: {failed_list}')
else:
    print('所有被试转换成功！')
```
